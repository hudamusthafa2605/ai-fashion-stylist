```python
def recommend_similar_products(product_index, n=5):
    similarity_scores = similarity_matrix[product_index]

    similar_indices = similarity_scores.argsort()[::-1]

    similar_indices = similar_indices[similar_indices != product_index]

    return data.iloc[similar_indices[:n]]
```

```python
def recommend_by_name(product_name, n=5):
    matches = data[data["product_name"].str.contains(
        product_name, case=False, na=False
    )]

    if matches.empty:
        return "Product not found"

    product_index = matches.index[0]

    return recommend_similar_products(product_index, n)
```

```python
def recommend_by_preferences(
    gender=None,
    occasion=None,
    season=None,
    color=None,
    max_price=None,
    n=5
):
    candidates = data.copy()

    if gender:
        candidates = candidates[
            candidates["Gender"].str.lower() == gender.lower()
        ]

    if occasion:
        candidates = candidates[
            candidates["occasion"].str.lower() == occasion.lower()
        ]

    if season:
        candidates = candidates[
            candidates["season"].str.lower() == season.lower()
        ]

    if color:
        candidates = candidates[
            candidates["color"].str.lower() == color.lower()
        ]

    if max_price:
        candidates = candidates[
            candidates["price_pkr"] <= max_price
        ]

    if candidates.empty:
        return "No matching products found."

    return candidates[
        ["product_name", "main_category", "color",
         "occasion", "season", "price_pkr"]
    ].head(n)
```

```python
def create_preference_vector(
    gender=None,
    occasion=None,
    season=None,
    color=None,
    fit_type=None,
    category_style=None,
    fabric=None
):
    preference = pd.DataFrame([{
        "Gender": gender,
        "occasion": occasion,
        "season": season,
        "color": color,
        "fit_type": fit_type,
        "Category Style": category_style,
        "fabric": fabric,
        "main_category": None
    }])

    return preference
```

```python
def encode_user_preferences(
    gender=None,
    occasion=None,
    season=None,
    color=None,
    fit_type=None,
    category_style=None,
    fabric=None,
    main_category=None
):
    user_preferences = pd.DataFrame([{
        "main_category": main_category,
        "color": color,
        "occasion": occasion,
        "season": season,
        "Gender": gender,
        "fit_type": fit_type,
        "Category Style": category_style,
        "fabric": fabric
    }])

    return encoder.transform(user_preferences)
```

```python
def recommend_outfits(
    gender=None,
    occasion=None,
    season=None,
    color=None,
    category_style=None,
    fabric=None,
    fit_type=None,
    main_category=None,
    max_price=None,
    n=5
):
    user_vector = encode_user_preferences(
        gender=gender,
        occasion=occasion,
        season=season,
        color=color
    )

    scores = cosine_similarity(user_vector, encoded_features)[0]

    candidates = data.copy()
    candidates["similarity_score"] = scores

    if max_price is not None:
        candidates = candidates[
            candidates["price_pkr"] <= max_price
        ]

    candidates = candidates.sort_values(
        "similarity_score",
        ascending=False
    )

    return candidates[
        ["product_name", "main_category", "color",
         "occasion", "season", "Gender",
         "price_pkr", "similarity_score"]
    ].head(n)
```

```python
def recommend_outfits(
    gender=None,
    occasion=None,
    season=None,
    color=None,
    category_style=None,
    fabric=None,
    fit_type=None,
    main_category=None,
    max_price=None,
    n=5
):
    user_vector = encode_user_preferences(
        gender=gender,
        occasion=occasion,
        season=season,
        color=color,
        category_style=category_style,
        fabric=fabric,
        fit_type=fit_type,
        main_category=main_category
    )

    scores = cosine_similarity(
        user_vector,
        encoded_features
    )[0]

    candidates = data.copy()
    candidates["similarity_score"] = scores

    if max_price is not None:
        candidates = candidates[
            candidates["price_pkr"] <= max_price
        ]

    candidates = candidates.sort_values(
        "similarity_score",
        ascending=False
    )

    candidates["match_percentage"] = (
        candidates["similarity_score"] * 100
    ).round(1)

    return candidates[
        [
            "product_name",
            "main_category",
            "color",
            "occasion",
            "season",
            "Gender",
            "Category Style",
            "fabric",
            "fit_type",
            "price_pkr",
            "match_percentage"
        ]
    ].head(n)
```
