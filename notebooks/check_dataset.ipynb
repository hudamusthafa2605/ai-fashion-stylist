```python
from sklearn.preprocessing import OneHotEncoder
import joblib
import pandas as pd
```

```python
data = pd.read_csv("../data/cleaned_fashion_data.csv")

features = data[[
    "color",
    "occasion",
    "season",
    "Gender",
    "fit_type",
    "fabric"
]].copy()

encoder = OneHotEncoder(handle_unknown="ignore")
encoded_features = encoder.fit_transform(features)

print(encoded_features.shape)
```

```python
joblib.dump(encoder, "../models/encoder.pkl")
joblib.dump(encoded_features, "../models/encoded_features.pkl")
```

```python
import pandas as pd
import joblib
from sklearn.metrics.pairwise import cosine_similarity

data = pd.read_csv("../data/cleaned_fashion_data.csv")

encoder = joblib.load("../models/encoder.pkl")
encoded_features = joblib.load("../models/encoded_features.pkl")

user_preferences = pd.DataFrame([{
    "color": "Trombone Yellow",
    "occasion": "Wedding/Formal",
    "season": "Summer",
    "Gender": "Male",
    "fit_type": "Free Size",
    "fabric": "Cotton"
}])

user_vector = encoder.transform(user_preferences)

scores = cosine_similarity(
    user_vector,
    encoded_features
)[0]

print(scores.max())
print(scores[:10])
```
